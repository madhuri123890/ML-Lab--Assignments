## Step 3: Word Frequency Vector (Bag-of-Words)

```python
def bag_of_words_vector(tokens, vocabulary):
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])
    return vector

vec = bag_of_words_vector(tokens, vocabulary)

print(vocabulary)
print(vec)
```

## Step 4: Comparing Two Sentences (Cosine Similarity)

```python
sentence_a = "machine learning uses numpy arrays"
sentence_b = "deep learning also uses numpy arrays heavily"

tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()

combined_vocab = np.unique(np.array(tokens_a + tokens_b))

vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)

print("Vector A:", vec_a)
print("Vector B:", vec_b)

cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))

print("Cosine similarity:", cosine_sim)
```

### Student Practice

```python
# Similar sentences
sentence1 = "I like machine learning"
sentence2 = "I enjoy machine learning"

# Very different sentences
sentence3 = "I like machine learning"
sentence4 = "The weather is very hot today"

tokens1 = sentence1.lower().split()
tokens2a = sentence2.lower().split()
tokens3 = sentence3.lower().split()
tokens4 = sentence4.lower().split()

# Similar pair
vocab_similar = np.unique(np.array(tokens1 + tokens2a))

vec1 = bag_of_words_vector(tokens1, vocab_similar)
vec2a = bag_of_words_vector(tokens2a, vocab_similar)

similarity1 = np.dot(vec1, vec2a) / (np.linalg.norm(vec1) * np.linalg.norm(vec2a))

# Different pair
vocab_different = np.unique(np.array(tokens3 + tokens4))

vec3 = bag_of_words_vector(tokens3, vocab_different)
vec4 = bag_of_words_vector(tokens4, vocab_different)

similarity2 = np.dot(vec3, vec4) / (np.linalg.norm(vec3) * np.linalg.norm(vec4))

print("Similarity of similar sentences:", similarity1)
print("Similarity of different sentences:", similarity2)
```

# Mini Project: Simple Text Similarity Checker (NLP Track)

```python
sentence1 = "I like machine learning"
sentence2 = "I enjoy machine learning"
sentence3 = "I like playing cricket"

sentences = [sentence1, sentence2, sentence3]

# Tokenize
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

# Shared vocabulary
shared_vocab = np.unique(
    np.array(tokens1 + tokens2 + tokens3)
)

# Bag-of-words vectors
vec1 = bag_of_words_vector(tokens1, shared_vocab)
vec2 = bag_of_words_vector(tokens2, shared_vocab)
vec3 = bag_of_words_vector(tokens3, shared_vocab)

# Cosine similarity function
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Calculate 3 pairs
sim_1_2 = cosine_similarity(vec1, vec2)
sim_1_3 = cosine_similarity(vec1, vec3)
sim_2_3 = cosine_similarity(vec2, vec3)

print("Similarity between Sentence 1 and Sentence 2:", sim_1_2)
print("Similarity between Sentence 1 and Sentence 3:", sim_1_3)
print("Similarity between Sentence 2 and Sentence 3:", sim_2_3)

# Find most similar pair
similarities = {
    "Sentence 1 and Sentence 2": sim_1_2,
    "Sentence 1 and Sentence 3": sim_1_3,
    "Sentence 2 and Sentence 3": sim_2_3
}

most_similar = max(similarities, key=similarities.get)

print("\nMost similar pair:", most_similar)
```
