Column Selection

```python
import pandas as pd

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    ),
    'two': pd.Series(
        [1, 2, 3, 4, 5, 6, 7, 8],
        index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
    )
}

d1 = pd.DataFrame(info)

print(d1['one'])
print(d1['two'])
```

Column Addition

```python
import pandas as pd

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5],
        index=['a', 'b', 'c', 'd', 'e']
    ),
    'two': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

df = pd.DataFrame(info)

print(df)

# Add new column by passing series
print("Add new column by passing series")

df['three'] = pd.Series(
    [20, 40, 60],
    index=['a', 'b', 'c']
)

print(df)

# Add new column using existing DataFrame columns
print("Add new column using existing DataFrame columns")

df['four'] = df['one'] + df['three']

print(df)
```

Column Deletion

```python
import pandas as pd

info = {
    'one': pd.Series([1, 2], index=['a', 'b']),
    'two': pd.Series([1, 2, 3], index=['a', 'b', 'c'])
}

df = pd.DataFrame(info)

print("The DataFrame:")
print(df)

# using del function
print("Delete the first column:")

del df['one']

print(df)

# using pop function
print("Delete the another column:")

df.pop('two')

print(df)
```

Selection by Label — loc

```python
import pandas as pd

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5],
        index=['a', 'b', 'c', 'd', 'e']
    ),
    'two': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

df = pd.DataFrame(info)

print("Data frame:")
print(df)

print("b index elements are:")
print(df.loc['b'])

print("e index elements are:")
print(df.loc['e'])
```

Selection by Integer Location — iloc

```python
import pandas as pd

info = {
    'one': pd.Series(
        [1, 2, 3, 4, 5],
        index=['a', 'b', 'c', 'd', 'e']
    ),
    'two': pd.Series(
        [1, 2, 3, 4, 5, 6],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

df = pd.DataFrame(info)

print(df)

print(df.iloc[3])
```

Addition of Rows

```python
import pandas as pd

d = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

d2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

# Replace append with concat
d = pd.concat([d, d2])

print(d)
```

Deletion of Rows

```python
import pandas as pd

a_info = pd.DataFrame(
    [[4, 5], [6, 7]],
    columns=['x', 'y']
)

b_info = pd.DataFrame(
    [[8, 9], [10, 11]],
    columns=['x', 'y']
)

a_info = pd.concat([a_info, b_info])

print(a_info)

# Drop rows with label 0
a_info = a_info.drop(0)

print(a_info)
```

DataFrame Column and Element Access

```python
import pandas as pd

data = {
    'Gender': ['F', 'M', 'M'],
    'Emp_ID': ['E01', 'E02', 'E03'],
    'Age': [25, 27, 25]
}

df = pd.DataFrame(
    data,
    columns=['Emp_ID', 'Gender', 'Age']
)

print(df)          # output all data
print(df.Emp_ID)   # output all Emp_ID
print(df.Emp_ID[0]) # output first Emp_ID
```

Removing Duplicates

```python
import pandas as pd

emp = {
    "Name": ["Siva", "Rama", "Krishna", "Siva"],
    "Age": [21, 32, 29, 21]
}

info = pd.DataFrame(emp)

print(info)

info = info.drop_duplicates()

print("After removing the duplicates the dataframe is :")
print(info)
```

Null Values Identification

```python
import pandas as pd
import numpy as np

info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

print(info)
print(info.count())

print("null values")
print(info.isnull())
print(info.isnull().sum())
```

Sorting Order

```python
import pandas as pd
import numpy as np

info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27., 29, np.nan, 32]
})

print(info)

print("Sorting Order")
print(info.sort_index())

print("Descending Order")
print(info.sort_index(ascending=False))
```
